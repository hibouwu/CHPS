## 1. Fundamental concept: regurarity / smoothness

One of the fundamental hypotheses in Machine Learning is **the regularity of the learning problem in the domain of the input data**. For a given model $f(\cdot)$ we could write in abstract terms that 

$$ ||f(x_1) - f(x_2)|| \ \sim \ ||x_1 - x_2|| $$  

which tells us that *the more close to each other are two data objects, $x_1$ and $x_2$, the more similar the respective outputs of our model are expected to be*. In both sides we have considered suitable norms (i.e. generally functions) that measure the distance between two models or data objects.

## 2. [Nearest Neighbors Classification](https://scikit-learn.org/stable/modules/neighbors.html#nearest-neighbors-classification)

### 2.1. Standard algorithm

Neighbors-based classification is a type of *instance-based learning* or *non-generalizing learning*: it does not attempt to construct a general internal model, but simply stores instances of the training data. Classification is typically computed from a simple *majority vote* of the nearest neighbors of each point: a query point is assigned the data class which has the most representatives within the nearest neighbors of the point.

Scikit-learn implements two different nearest neighbors classifiers: 
* ``KNeighborsClassifier`` implements learning based on the $k$ nearest neighbors of each query point, where $k$ is an integer value specified by the user. This is the most commonly used technique. The optimal choice of the value $k$ is highly data-dependent: in general a larger $k$ suppresses the effects of noise, but makes the classification boundaries less distinct.
* ``RadiusNeighborsClassifier`` implements learning based on the number of neighbors within a fixed radius  of each training point, where $r$ is a floating-point value specified by the user. In cases where the data is not uniformly sampled, radius-based neighbors classification in RadiusNeighborsClassifier can be a better choice. The user specifies a fixed radius $r$, such that points in sparser neighborhoods use fewer nearest neighbors for the classification. For high-dimensional parameter spaces, this method becomes less effective due to the so-called “curse of dimensionality”.

**Vote weighting schemes:** The basic nearest neighbors classification uses uniform weights: that is, the value assigned to a query point is computed from a simple majority vote of the nearest neighbors. Under some circumstances, it is better to weight the neighbors such that nearer neighbors contribute more to the fit. This can be accomplished through the ``weights`` keyword. The default value, ``weights = 'uniform'``, assigns uniform weights to each neighbor. ``weights = 'distance'`` assigns weights proportional to the inverse of the distance from the query point. Alternatively, a user-defined function of the distance can be supplied to compute the weights.

**Neighbor search:** As one can see, this type of algorithm can be computationally heavy for large datasets. The core operation is what is called *neighbor search* (note: this is a scientific problem on its own), which is to find the neighbors of a given point out of the whole dataset. The brute force algorithm goes through all data objects one by one and is only applicable in small datasets. More efficient search uses sophisticated data structures (e.g. k-d trees, ball trees...), and can provide either approximate or accurate results.

**Application on the Iris dataset**

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from sklearn import neighbors, datasets

k = 5

# import some data to play with
iris = datasets.load_iris()

# just as a toy case, we only take two features
X = iris.data[:, 2:4]
y = iris.target

h = .02  # step size in the mesh

# Create color maps
cmap_light = ListedColormap(['#FFAAAA', '#AAFFAA', '#AAAAFF'])
cmap_bold  = ListedColormap(['#FF0000', '#00FF00', '#0000FF'])

# apply the two vote weighting schemens
for weights in ['uniform', 'distance']:
    # we create an instance of Neighbours Classifier and fit the data.
    clf = neighbors.KNeighborsClassifier(k, weights=weights)
    clf.fit(X, y)

    # Plot the decision boundary. For that, we will assign a color to each
    # point in the mesh [x_min, x_max] x [y_min, y_max].
    x_min, x_max = X[:, 0].min() - 0.2, X[:, 0].max() + 0.2
    y_min, y_max = X[:, 1].min() - 0.2, X[:, 1].max() + 0.2
    xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                         np.arange(y_min, y_max, h))
    Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])

    # Put the result into a color plot
    Z = Z.reshape(xx.shape)
    plt.figure()
    plt.pcolormesh(xx, yy, Z, cmap=cmap_light, shading='auto')

    # Plot also the training points
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap=cmap_bold,
                edgecolor='k', s=20)
    plt.xlim(xx.min(), xx.max())
    plt.ylim(yy.min(), yy.max())
    plt.title("3-Class classification (k = %i, weights = '%s')"
              % (k, weights))

plt.show()
````

### 2.2. Nearest Centroid Classifier

The ``NearestCentroid`` classifier is a simple algorithm that represents
each class by the centroid of its members. In effect, this makes it
similar to the label updating phase of the $k$-means algorithm.
It also has no parameters to choose, making it a good baseline classifier. It
does, however, suffer when classes have non-convex shapes, as well as when classes have
very different variances, because what it assumes is equal variance in all dimensions. See Linear Discriminant Analysis (``sklearn.discriminant_analysis.LinearDiscriminantAnalysis``)
and Quadratic Discriminant Analysis (``sklearn.discriminant_analysis.QuadraticDiscriminantAnalysis``)
for more complex methods that do not make this assumption.

Note that here there is no notion of $k$ as number of neighbors...

**Simple example of the default ``NearestCentroid`` method**

````python
import numpy as np
from sklearn.neighbors import NearestCentroid

# create some 2D points
X = np.array([[-1, -1], [-2, -1], [-3, -2], [1, 1], [2, 1], [3, 2]])
# set the ground truth labels
y = np.array([1, 1, 1, 2, 2, 2])

# create and fit the model (i.e. compute the centroids of the data classes)
clf = NearestCentroid()
clf.fit(X, y)

# predict for a new observation
point = [-0.8, -1]
print(clf.predict([point]))
````

**Nearest Shrunken Centroid**

The ``NearestCentroid`` classifier has a ``shrink_threshold`` parameter, which implements the nearest shrunken centroid classifier. In effect, the value of each feature for each centroid is divided by the within-class variance of that feature. The feature values are then reduced by ``shrink_threshold``. Most notably, if a particular feature value crosses zero, it is set
to zero. This removes the feature from affecting the classification and is useful, for example, for removing noisy features.

In the example below, using a small shrink threshold increases the accuracy of
the model from 0.81 to 0.82.

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
from sklearn import datasets
from sklearn.neighbors import NearestCentroid

# import some data and take the first two features
iris = datasets.load_iris()
X = iris.data[:, :2]
y = iris.target

h = .02  # step size in the mesh

# Create color maps
cmap_light = ListedColormap(['#FFAAAA', '#AAFFAA', '#AAAAFF'])
cmap_bold = ListedColormap(['#FF0000', '#00FF00', '#0000FF'])

for shrinkage in [None, .2]:
    # we create an instance of Neighbours Classifier and fit the data.
    clf = NearestCentroid(shrink_threshold=shrinkage)
    clf.fit(X, y)
    y_pred = clf.predict(X)
    print(shrinkage, np.mean(y == y_pred))
    # Plot the decision boundary. For that, we will assign a color to each
    # point in the mesh [x_min, x_max]x[y_min, y_max].
    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                         np.arange(y_min, y_max, h))
    Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])

    # Put the result into a color plot
    Z = Z.reshape(xx.shape)
    plt.figure()
    plt.pcolormesh(xx, yy, Z, cmap=cmap_light, shading='auto')

    # Plot also the training points
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap=cmap_bold,
                edgecolor='k', s=20)
    plt.title("3-Class classification (shrink_threshold=%r)"
              % shrinkage)
    plt.axis('tight')

plt.show()
````

## 3. Decision Trees

### 3.1. [Simple example on Iris dataset](https://scikit-learn.org/stable/auto_examples/tree/plot_iris_dtc.html#sphx-glr-auto-examples-tree-plot-iris-dtc-py)

Decision surface of a decision tree trained on pairs of features of the iris dataset. For each pair of data features, the decision tree learns *decision boundaries* using combinations of *simple thresholding rules* that are inferred from the training samples.

The tree structure of a tree built on all of the features is also shown.

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree

# Parameters
n_classes = 3
plot_colors = "ryb"
plot_step = 0.02

plt.figure(figsize=(13,7))

# Load data
iris = load_iris()
# go through the pairs of features (there are 4 features in total)
for pairidx, pair in enumerate([[0, 1], [0, 2], [0, 3], [1, 2], [1, 3], [2, 3]]):
    # Take the two features
    X = iris.data[:, pair]
    y = iris.target

    # Train the decision tree
    clf = DecisionTreeClassifier().fit(X, y)

    # Plot the decision boundary
    
    plt.subplot(2, 3, pairidx + 1)

    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, plot_step),
                         np.arange(y_min, y_max, plot_step))
    plt.tight_layout(h_pad=0.5, w_pad=0.5, pad=2.5)

    Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    cs = plt.contourf(xx, yy, Z, cmap=plt.cm.RdYlBu)

    plt.xlabel(iris.feature_names[pair[0]])
    plt.ylabel(iris.feature_names[pair[1]])

    # Plot the training points
    for i, color in zip(range(n_classes), plot_colors):
        idx = np.where(y == i)
        plt.scatter(X[idx, 0], X[idx, 1], c=color, label=iris.target_names[i],
                    cmap=plt.cm.RdYlBu, edgecolor='black', s=15)

plt.suptitle("Decision surface of a decision tree using paired features")
plt.legend(loc='lower right', borderpad=0, handletextpad=0)
plt.axis("tight")

# creates a tree object and shows the default parametrization
dtree = DecisionTreeClassifier()
# fits the tree
dtree.fit(iris.data, iris.target)

plt.figure(figsize=(20,20))

# This visualizes the tree. You should use the latest scikit-learn version.
# Recall that you can do that by running in a code cell: !pip install -U scikit-learn 
tree.plot_tree(dtree, filled=True)
plt.show()
````

The **Gini coefficient** reported at each tree node is an index in $[0, 1-1/K]$ that measures the inequality in a probability distribution. In this context, it measures the extent to which each node gathers samples of one only data class. So, note that when there is a perfect purity, e.g. $Gini([0,47,0]) = 0$, $Gini([50,50,0]) = 0.5$, and $Gini([50,50,50]) = 0.667$.

### 3.2. How the tree is computed

A tree is grown **greedily** and **top-down**: at each node we look at the samples that reached it, pick the split that best separates their classes, send the samples to the children, and repeat. A branch stops growing when a stopping criterion is met, and the node becomes a leaf that predicts the majority class of its samples.

```
----------------------------------------
GrowTree(S, F, SplitCriterion, StoppingCriterion):   # S: samples at the node, F: available features
    if StoppingCriterion(S) :                        # e.g. node is pure, too few samples, max depth reached
        return Leaf( class distribution of S )       # what to output (majority class, class probabilities,
                                                     # cost-sensitive decision...) is decided at prediction time

    best = None
    for each feature f in F :
        for each candidate split s of S on f :       # all values of f (categorical), or thresholds f < t (numerical)
            if SplitCriterion(S, s) is better than best :
                best = s

    if best gives no improvement :                   # second stopping rule
        return Leaf( class distribution of S )

    node = Node( test = best )
    for each subset S_v produced by best :
        node.child[v] = GrowTree(S_v, F', SplitCriterion, StoppingCriterion)   # F'=F\f (C4.5), or =F (CART: f can be reused)
    return node

Tree = Prune( GrowTree(S, F, SplitCriterion, StoppingCriterion) )              # post-pruning, to reduce overfitting
----------------------------------------
```

The two criteria in capitals are what makes one tree algorithm differ from another:

* **CART** (scikit-learn) uses the **Gini impurity**, $\text{Gini}(S) = 1 - \sum_c p_c^2$, splits are always **binary**, and features can be reused along a branch.
* **C4.5** (and its ancestor ID3) uses the **entropy** $H(S)$ and selects the split with the highest **InfoGain**, with one branch per value of a categorical feature.

$$H(S) = -\sum_{\text{class}\ c} p_c \log_2 p_c, \ \ \ \ p_c = \frac{\text{number of elements of class}\ c\ \text{in set}\ S}{|S|}$$  

$$\text{InfoGain}(S, f) = H(S) - \sum_{\text{branch}\ v} \pi_v\, H(S_v), \ \ \ \ \pi_v = \frac{|S_v|}{|S|}$$

Since InfoGain favors features with many values (in the extreme, an Id feature splits every sample into its own pure branch), C4.5 normalizes it by the entropy of the split itself and uses the **information gain ratio**:

$$\text{InfoGainRatio}(S, f) = \frac{\text{InfoGain}(S, f)}{\text{SplitInfo}(S, f)}, \qquad \text{SplitInfo}(S, f) = -\sum_{\text{branch}\ v} \pi_v \, \log_2 \pi_v$$

Entropy is over the class proportions inside a node; SplitInfo is over the branch proportions of the split, i.e. the entropy of the partition induced by $f$, ignoring the labels entirely.

Note that in scikit-learn you can also ask for the entropy criterion (``DecisionTreeClassifier(criterion="entropy")``), but the splits remain binary and there is no gain ratio.

**A step-by-step example (C4.5)**

A toy dataset (Quinlan's classic): 14 days described by 4 categorical features, and the target "Play".

| Id | Outlook | Temperature | Humidity | Wind | Play |
|:---:|:---|:---|:---|:---|:---:|
| 1 | Sunny | Hot | High | Weak | No |
| 2 | Sunny | Hot | High | Strong | No |
| 3 | Overcast | Hot | High | Weak | Yes |
| 4 | Rain | Mild | High | Weak | Yes |
| 5 | Rain | Cool | Normal | Weak | Yes |
| 6 | Rain | Cool | Normal | Strong | No |
| 7 | Overcast | Cool | Normal | Strong | Yes |
| 8 | Sunny | Mild | High | Weak | No |
| 9 | Sunny | Cool | Normal | Weak | Yes |
| 10 | Rain | Mild | Normal | Weak | Yes |
| 11 | Sunny | Mild | Normal | Strong | Yes |
| 12 | Overcast | Mild | High | Strong | Yes |
| 13 | Overcast | Hot | Normal | Weak | Yes |
| 14 | Rain | Mild | High | Strong | No |

**Root node.** The 14 samples are 9 Yes and 5 No, so $H(S) = -\frac{9}{14}\log_2\frac{9}{14} - \frac{5}{14}\log_2\frac{5}{14} = 0.940$. We try each feature:

| Feature | InfoGain | SplitInfo | GainRatio |
|:---|:---:|:---:|:---:|
| Outlook | **0.247** | 1.577 | **0.156** |
| Humidity | 0.152 | 1.000 | 0.152 |
| Wind | 0.048 | 0.985 | 0.049 |
| Temperature | 0.029 | 1.557 | 0.019 |

Outlook is selected. It gives three branches: Sunny (2 Yes, 3 No, $H=0.971$), **Overcast (4 Yes, 0 No, $H=0$)**, and Rain (3 Yes, 2 No, $H=0.971$). The Overcast branch is pure, so it becomes a leaf "Yes" and that branch stops.

**Sunny branch** (samples 1, 2, 8, 9, 11), where we now choose among the remaining features:

| Feature | InfoGain | GainRatio |
|:---|:---:|:---:|
| Humidity | **0.971** | **1.000** |
| Temperature | 0.571 | 0.375 |
| Wind | 0.020 | 0.021 |

Humidity separates the branch perfectly: High gives 3 No, Normal gives 2 Yes. Both children are pure leaves.

**Rain branch** (samples 4, 5, 6, 10, 14):

| Feature | InfoGain | GainRatio |
|:---|:---:|:---:|
| Wind | **0.971** | **1.000** |
| Temperature | 0.020 | 0.021 |
| Humidity | 0.020 | 0.021 |

Wind separates it perfectly: Weak gives 3 Yes, Strong gives 2 No.

The learned tree is the following (the content of each node is written as in the trees that scikit-learn draws: the test, the impurity of the node, the number of samples that reached it, their class counts `value = [No, Yes]`, and the majority class):

```
                            ┌─────────────────┐
                            │ Outlook?        │
                            │ entropy = 0.940 │
                            │ samples = 14    │
                            │ value = [5, 9]  │
                            │ class = Yes     │
                            └─────────────────┘
                                     │
         ┌───────────────────────┬───┴───────────────────┐
   Sunny │              Overcast │                  Rain │
         │                       │                       │
┌─────────────────┐     ┌─────────────────┐     ┌─────────────────┐
│ Humidity?       │     │ leaf            │     │ Wind?           │
│ entropy = 0.971 │     │ entropy = 0.000 │     │ entropy = 0.971 │
│ samples = 5     │     │ samples = 4     │     │ samples = 5     │
│ value = [3, 2]  │     │ value = [0, 4]  │     │ value = [2, 3]  │
│ class = No      │     │ class = Yes     │     │ class = Yes     │
└─────────────────┘     └─────────────────┘     └─────────────────┘
         │                                               │
         ┌─────────────────────┐                     ┌───┴─────────────────┐
    High │              Normal │                Weak │              Strong │
         │                     │                     │                     │
┌─────────────────┐   ┌─────────────────┐   ┌─────────────────┐   ┌─────────────────┐
│ leaf            │   │ leaf            │   │ leaf            │   │ leaf            │
│ entropy = 0.000 │   │ entropy = 0.000 │   │ entropy = 0.000 │   │ entropy = 0.000 │
│ samples = 3     │   │ samples = 2     │   │ samples = 3     │   │ samples = 2     │
│ value = [3, 0]  │   │ value = [0, 2]  │   │ value = [0, 3]  │   │ value = [2, 0]  │
│ class = No      │   │ class = Yes     │   │ class = Yes     │   │ class = No      │
└─────────────────┘   └─────────────────┘   └─────────────────┘   └─────────────────┘
```

Note that Temperature is never used: by choosing splits, a tree also performs feature selection. Note also what C4.5 adds beyond this sketch: numerical features are handled by searching the best threshold $f < t$ (as in the Iris example of Section 3.1), missing values are handled by sending a sample to all branches with fractional weights, and the fully grown tree is pruned afterwards.

**What CART would do on the same data**

The first split above is **not binary** because C4.5 splits a categorical feature multiway, with one branch per value. CART (hence scikit-learn) always splits in two, so on the same data it asks a yes/no question. Here, its first split is **"Outlook = Overcast?"**: the Yes side is the same pure group of 4 days, but the No side keeps the remaining 10 days (5 Yes, 5 No) together and has to be split again further down, possibly testing Outlook a second time.

| | C4.5 | CART (scikit-learn) |
|:---|:---|:---|
| split of a categorical feature | one branch per value | binary (one value, or a subset, against the rest) |
| split of a numerical feature | binary, $f < t$ | binary, $f \leq t$ |
| criterion | InfoGain / InfoGainRatio (entropy) | Gini decrease (or entropy) |
| feature reuse along a branch | no, for a categorical feature | yes |
| resulting tree on our 14 days | depth 2, 6 leaves, Temperature unused | depth 4, 7 leaves, all features used |

The cell below reproduces it: scikit-learn has no native support for categorical features, so the four features are first one-hot encoded, which is also what forces the binary questions.

````python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, export_text, plot_tree

rows = [("Sunny","Hot","High","Weak","No"),      ("Sunny","Hot","High","Strong","No"),
        ("Overcast","Hot","High","Weak","Yes"),  ("Rain","Mild","High","Weak","Yes"),
        ("Rain","Cool","Normal","Weak","Yes"),   ("Rain","Cool","Normal","Strong","No"),
        ("Overcast","Cool","Normal","Strong","Yes"), ("Sunny","Mild","High","Weak","No"),
        ("Sunny","Cool","Normal","Weak","Yes"),  ("Rain","Mild","Normal","Weak","Yes"),
        ("Sunny","Mild","Normal","Strong","Yes"),("Overcast","Mild","High","Strong","Yes"),
        ("Overcast","Hot","Normal","Weak","Yes"),("Rain","Mild","High","Strong","No")]
df = pd.DataFrame(rows, columns=["Outlook", "Temperature", "Humidity", "Wind", "Play"])

# one-hot encoding of the categorical features (scikit-learn does not handle them natively)
X = pd.get_dummies(df.drop(columns="Play"))
y = (df["Play"] == "Yes").astype(int)

clf = DecisionTreeClassifier(criterion="entropy", random_state=0).fit(X, y)
print("depth =", clf.get_depth(), "  leaves =", clf.get_n_leaves())
print(export_text(clf, feature_names=list(X.columns)))

# the same tree, drawn as in Section 3.1
plt.figure(figsize=(16, 8))
plot_tree(clf, feature_names=list(X.columns), class_names=["No", "Yes"],
          filled=True, rounded=True, fontsize=9)
plt.show()
````

### 3.3. Playing with the impurity measures

The cell below plots the two impurity measures for a node with two classes, as a function of the proportion $p$ of the positive class in the node, and marks the value at the $p$ you select with the slider. Both are maximal for the most mixed node ($p=0.5$: Gini $=0.5$, entropy $=1$ bit) and zero for a pure node.

Then you can set the class counts of the two children of a split, and see the resulting InfoGain and Gini decrease.

````python
import numpy as np
import matplotlib.pyplot as plt

def gini(p):
    """Gini impurity of a node with a proportion p of positive samples."""
    return 1 - p**2 - (1-p)**2

def entropy(p):
    """Entropy (in bits) of a node with a proportion p of positive samples."""
    p = np.asarray(p, dtype=float)
    with np.errstate(divide='ignore', invalid='ignore'):
        e = -(p*np.log2(p) + (1-p)*np.log2(1-p))
    return np.where((p == 0) | (p == 1), 0.0, e)

def impurity_plot(p=0.5):
    ps = np.linspace(0, 1, 201)
    plt.figure(figsize=(7, 4))
    plt.plot(ps, gini(ps), label=r'Gini $= 1-p^2-(1-p)^2$')
    plt.plot(ps, entropy(ps), label=r'Entropy $= -p\log_2 p-(1-p)\log_2(1-p)$')
    plt.scatter([p, p], [gini(p), entropy(p)], color='k', zorder=5)
    plt.axvline(p, color='k', lw=0.8, ls=':')
    plt.xlabel('$p$ = proportion of the positive class in the node')
    plt.ylabel('impurity')
    plt.legend(loc='lower center', fontsize='small')
    plt.title('p = %.2f    Gini = %.3f    Entropy = %.3f' % (p, gini(p), entropy(p)))
    plt.tight_layout()
    plt.show()

def split_gains(left_pos=3, left_neg=0, right_pos=1, right_neg=4):
    """InfoGain and Gini decrease of a binary split, given the class counts of the two children."""
    n_l, n_r = left_pos + left_neg, right_pos + right_neg
    n = n_l + n_r
    p_par = (left_pos + right_pos) / n
    p_l = left_pos / n_l if n_l else 0
    p_r = right_pos / n_r if n_r else 0
    H = lambda p: float(entropy(p))
    G = lambda p: float(gini(p))
    infogain = H(p_par) - (n_l/n)*H(p_l) - (n_r/n)*H(p_r)
    ginidrop = G(p_par) - (n_l/n)*G(p_l) - (n_r/n)*G(p_r)
    print('parent: %2d pos / %2d neg    Entropy = %.3f   Gini = %.3f' % (left_pos+right_pos, left_neg+right_neg, H(p_par), G(p_par)))
    print('left  : %2d pos / %2d neg    Entropy = %.3f   Gini = %.3f' % (left_pos, left_neg, H(p_l), G(p_l)))
    print('right : %2d pos / %2d neg    Entropy = %.3f   Gini = %.3f' % (right_pos, right_neg, H(p_r), G(p_r)))
    print('--> InfoGain = %.3f     Gini decrease = %.3f' % (infogain, ginidrop))

try:
    # interactive version (ipywidgets comes with Anaconda)
    from ipywidgets import interact, FloatSlider, IntSlider
    interact(impurity_plot, p=FloatSlider(min=0, max=1, step=0.01, value=0.5))
    interact(split_gains,
             left_pos=IntSlider(min=0, max=10, value=3), left_neg=IntSlider(min=0, max=10, value=0),
             right_pos=IntSlider(min=0, max=10, value=1), right_neg=IntSlider(min=0, max=10, value=4))
except ImportError:
    # static fallback: just call the functions with your own values
    impurity_plot(0.5)
    split_gains(3, 0, 1, 4)
````

## 4. [Support Vector Machines](https://scikit-learn.org/stable/modules/svm.html#classification)

Support vector machines (SVMs) are a family of supervised learning methods used for classification, regression and outliers detection.

### Classification with Support Vector Machines (SVMs)

Some of the material comes from Chapter 12 of the book [Mathematics for Machine Learning](https://mml-book.github.io/book/mml-book.pdf).


**Starting point: deciding using a linear model**

<img src="multiple-hyperplanes1.png" alt="Drawing" style="width: 600px;"/>

Datapoints: $x \in \mathbb{R}^d$

Binary label: $y \in \{-1,+1\}$

Linear function: $f(x) = w^\top x + b$ 

... alternative notation using the inner product $f(x) = \langle w, x \rangle + b$ 

<img src="hyperplane-eq.png" alt="Drawing" style="width: 600px;"/>

* $w^\top x_i + b = 0$ for point $x_i$ on the plane

* $w^\top x_i + b > 0$ for point $x_i$ on the positive side of the plane ($y_i=+1$)

* $w^\top x_i + b < 0$ for point $x_i$ on the negative side of the plane ($y_i=-1$)

Convention: take the plane as part of the positive side, and arrive to a **simple linear decision function** (combines the above three)

$$ y_i (w^\top x_i + b) \geq 0 $$

**Requiring a large strict margin**

Interesting to note that the relation of a given point $x$, the orthogonal projection point $x'$ on the hyperplane, and their distance $r = \lVert x - x' \rVert$ can be given by 
 
$$ x = x' + r \frac{w}{\lVert w \rVert}$$

where generally we assume for convenience that the Euclidean norm $\lVert w \rVert = \sqrt{w^\top w}$.

In words this expression says that: starting from $x'$, if we follow the direction of the unit-length vector $\frac{w}{\lVert w \rVert}$ for a distance $r$, we can arrive to $x$.  

<img src="r-distance.png" alt="Drawing" style="width: 250px;"/>

The new objective requiring the two data classes to be at least $r$ distance from the hyperplane is

$$ y_i (w^\top x_i + b) \geq r $$

In theory, *the larger the margin, the simpler the problem and therefore a less complex model is sufficient to treat it* (see Vapnik, The Nature of Statistical Learning Theory, Springer, 2000). 

Therefore a simple optimization that asks for maximum margin can be written as

$$ \max_{w,b,r} r $$

$$ \text{subject to }\ \ y_i (w^\top x_i + b) \geq r,\ \  \forall i, \ \ \text{and } \ \lVert w \rVert = 1, \ \ r>0 $$

Instead of looking for a solution where $w$ is a unit-length vector ($\lVert w \rVert = 1$), it is equivalent to define the margin as 

$$
r = \frac{1}{\lVert w\rVert}
$$

Notes: 

* The equivalence comes from the fact that either the margin will be a quantity that scales properly to the unscaled data, or alternatively, we can think of it as a unit length quantity refering to the properly scaled version of the data such as that the margin is $r=1$.

* Another point is that instead of using $\max_{w,b} \frac{1}{\lVert w\rVert}$ we can use $\min_{w,b} \frac{1}{2}\lVert w\rVert^2$, where the constant $\frac{1}{2}$ is added for convenience for when we compute gradients -- recall that $\lVert w\rVert^2 = w^\top w$.

Integrating the above points, we can express the minimization by basically removing the $r=1$ variable, as follows

$$ \min_{w,b} \frac{1}{2}\lVert w\rVert^2 $$

$$ \text{subject to }\ \ y_i (w^\top x_i + b) \geq 1, \ \ \forall i$$

This objective is "hard" because it estimates the margin by not allowing any violation of the side in which the points are: positives should be on the one side and the negatives on the other.

**Soft margin for non-linearly separable data**

<img src="multiple-hyperplanes2.png" alt="Drawing" style="width: 300px;"/>

The main idea is to introduce a *slack variable* $\xi_i$ for each datapoint $x_i$ that allows it to be within the margin or even on the wrong side of the hyperplane.

<img src="soft-margin.png" alt="Drawing" style="width: 370px;"/>

$$ \min_{w,b} \frac{1}{2}\lVert w\rVert^2 + C \sum_{i=1}^{n} \xi_i$$

$$ \text{subject to }\ \ y_i (w^\top x_i + b) \geq 1 - \xi_i, \ \ \text{and }\ \ \xi_i \geq 0, \ \ \forall i \ \ $$

This formulation is called $C$-SVM, and $C > 0$ is a user-defined parameter that expresses the trade-off between maximizing the margin and a large sum of slack terms for the points. Higher $C$ values make the margin to be smaller, therefore $C$ is inversely related to the eventual regularization.

Solving the optimization problem to fit the model is discussed in Sec.12.3.1 of the referred book.

**Multiclass SVM**

Apparently all the above discussion addresses the problem of separating specifically $2$ classes out of $K$ (same as for the Logistic Regression). There is no adaptation of the native problem for multiple classes, however there are strategies that can deal with multiple classes:

* By building $K$ **1-vs-all SVM classifiers**, one per class. This means that the classifier learns to separate a target class versus all the other data classes.

* By building $\frac{K(K-1)}{2}$ **1-vs-1 SVM classifiers**, one for every pair of classes. This means that each classifier is specialized in separating a pair of specific classes.

To put such approaches to work there is a need for defining ways to combine the decisions of those multiple classifiers for a new datapoint. 

### Discussion

The advantages of SVMs are:

* Effective in high dimensional spaces.
* Still effective in cases where the number of features is greater than the number of samples.
* Uses a subset of training points in the decision function (called *support vectors*), so it is also memory efficient.
* Versatile: different Kernel functions can be specified for the decision function. Common kernels are provided, but it is also possible to specify custom ones.

The disadvantages of SVMs include:

* If the number of features is much greater than the number of samples, then avoid over-fitting in choosing Kernel functions and regularization term is crucial.
* SVMs do not directly provide probability estimates, these are usually calculated using expensive cross-validation.
* The support vector machines in scikit-learn support both dense (numpy.ndarray and convertible to that by numpy.asarray) and sparse (any scipy.sparse) sample vectors as input. However, to use an SVM to make predictions for sparse data, it must have been fit on such data. For optimal performance, use C-ordered numpy.ndarray (dense) or scipy.sparse.csr_matrix (sparse) with dtype=float64.

**Online demo:** [https://dash-gallery.plotly.host/dash-svm/](https://dash-gallery.plotly.host/dash-svm/)

### 4.1 [Maximum margin separating hyperplane](https://scikit-learn.org/stable/auto_examples/svm/plot_separating_hyperplane.html#sphx-glr-auto-examples-svm-plot-separating-hyperplane-py)

Plot the maximum margin separating hyperplane within a two-class separable dataset using a Support Vector Machine classifier with linear kernel.

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import svm
from sklearn.datasets import make_blobs

# we create 40 separable points
X, y = make_blobs(n_samples=40, centers=2, random_state=6)

# fit the model, don't regularize for illustration purposes
clf = svm.SVC(kernel='linear', C=1000)
clf.fit(X, y)

plt.scatter(X[:, 0], X[:, 1], c=y, s=30, cmap=plt.cm.Paired)

# plot the decision function
ax = plt.gca()
xlim = ax.get_xlim()
ylim = ax.get_ylim()

# create grid to evaluate model
xx = np.linspace(xlim[0], xlim[1], 30)
yy = np.linspace(ylim[0], ylim[1], 30)
YY, XX = np.meshgrid(yy, xx)
xy = np.vstack([XX.ravel(), YY.ravel()]).T
Z = clf.decision_function(xy).reshape(XX.shape)

# plot decision boundary and margins
ax.contour(XX, YY, Z, colors='k', levels=[-1, 0, 1], alpha=0.5,
           linestyles=['--', '-', '--'])
# plot support vectors
ax.scatter(clf.support_vectors_[:, 0], clf.support_vectors_[:, 1], s=100,
           linewidth=1, facecolors='none', edgecolors='k')
plt.show()
````

### 4.2 [SVM with various kernels](https://scikit-learn.org/stable/auto_examples/exercises/plot_iris_exercise.html#sphx-glr-auto-examples-exercises-plot-iris-exercise-py)

A tutorial exercise for using different SVM kernels. 
This is used in this [supervised learning tutorial](https://scikit-learn.org/stable/tutorial/statistical_inference/supervised_learning.html#using-kernels-tut).

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets, svm

iris = datasets.load_iris()
X = iris.data
y = iris.target

X = X[y != 0, :2]
y = y[y != 0]

n_sample = len(X)

np.random.seed(0)
order = np.random.permutation(n_sample)
X = X[order]
y = y[order].astype(float)

X_train = X[:int(.9 * n_sample)]
y_train = y[:int(.9 * n_sample)]
X_test = X[int(.9 * n_sample):]
y_test = y[int(.9 * n_sample):]

# fit the model
for kernel in ('linear', 'rbf', 'poly'):
    clf = svm.SVC(kernel=kernel, gamma=10)
    clf.fit(X_train, y_train)

    plt.figure()
    plt.clf()
    plt.scatter(X[:, 0], X[:, 1], c=y, zorder=10, cmap=plt.cm.Paired,
                edgecolor='k', s=20)

    # Circle out the test data
    plt.scatter(X_test[:, 0], X_test[:, 1], s=80, facecolors='none',
                zorder=10, edgecolor='k')

    plt.axis('tight')
    x_min = X[:, 0].min()
    x_max = X[:, 0].max()
    y_min = X[:, 1].min()
    y_max = X[:, 1].max()

    XX, YY = np.mgrid[x_min:x_max:200j, y_min:y_max:200j]
    Z = clf.decision_function(np.c_[XX.ravel(), YY.ravel()])

    # Put the result into a color plot
    Z = Z.reshape(XX.shape)
    plt.pcolormesh(XX, YY, Z > 0, cmap=plt.cm.Paired, shading='auto')
    plt.contour(XX, YY, Z, colors=['k', 'k', 'k'],
                linestyles=['--', '-', '--'], levels=[-.5, 0, .5])

    plt.title(kernel)
plt.show()
````

### 4.3. Cross-validation

**SVM using a single data split**

Note that $C>0$ is a regularization parameter. The strength of the regularization is inversely proportional to $C$. In fact this is a squared $\ell_2$ penalty.

````python
from sklearn import datasets, svm
digits = datasets.load_digits()
X_digits = digits.data
y_digits = digits.target
svc = svm.SVC(C=1, kernel='linear')
svc.fit(X_digits[:-100], y_digits[:-100]).score(X_digits[-100:], y_digits[-100:])
````

**SVM using $k$-fold cross-validation**

````python
import numpy as np

X_folds = np.array_split(X_digits, 3)
y_folds = np.array_split(y_digits, 3)
scores = list()

# train and test for the split indicated by each fold
for k in range(3):
    # We use 'list' to copy, in order to 'pop' later on
    X_train = list(X_folds)
    X_test = X_train.pop(k)
    X_train = np.concatenate(X_train)
    y_train = list(y_folds)
    y_test = y_train.pop(k)
    y_train = np.concatenate(y_train)
    scores.append(svc.fit(X_train, y_train).score(X_test, y_test))

# show fitting performance for the different folds
print(scores)
````

#### [Cross-validation on Digits Dataset Exercise](https://scikit-learn.org/stable/auto_examples/exercises/plot_cv_digits.html#sphx-glr-auto-examples-exercises-plot-cv-digits-py)

A tutorial exercise using Cross-validation with an SVM on the Digits dataset.

This exercise is used in this [CV tutorial](https://scikit-learn.org/stable/tutorial/statistical_inference/model_selection.html#cv-generators-tut).

````python
import numpy as np
from sklearn.model_selection import cross_val_score
from sklearn import datasets, svm
import matplotlib.pyplot as plt

# load digits dataset
digits = datasets.load_digits()
X = digits.data
y = digits.target

# create the svm model/object
svc = svm.SVC(kernel='linear')
# set the search space for the model parameter C
C_s = np.logspace(-10, 0, 10)

scores = list()
scores_std = list()

# do the following for each different C value in the search space
for C in C_s:
    svc.C = C
    this_scores = cross_val_score(svc, X, y, cv=5, n_jobs=1)
    scores.append(np.mean(this_scores))
    scores_std.append(np.std(this_scores))

# Do the plotting
plt.figure()
plt.semilogx(C_s, scores)
plt.semilogx(C_s, np.array(scores) + np.array(scores_std), 'b--')
plt.semilogx(C_s, np.array(scores) - np.array(scores_std), 'b--')
locs, labels = plt.yticks()
plt.yticks(locs, list(map(lambda x: "%g" % x, locs)))
plt.ylabel('CV score')
plt.xlabel('Parameter C')
plt.ylim(0, 1.1)
plt.show()
````

## 5. Curse of Dimensionality

In all the related fields of Data Science, it is usual to hear that data in *high-dimensions are sparse and that the distance calculations inevitably become less trustworthy in such cases*.

**a)** Here is a simple illustrative example to show the effect of what is called *curse of dimensionality*. Suppose we have two points $a = [0]^{1\times d}$ and $b = [1]^{1\times d}$, and considering the Euclidean distance $||a-b||_2$ we can see:
* $d=1$: in 1D, (0) and (1) have distance equal to: $\sqrt{(0-1)^2} = 1$. 
* $d=2$: in 2D, (0,0) and (1,1) have distance $\sqrt{(0-1)^2+(0-1)^2} = \sqrt{2}$.
* $d=3$: in 3D, (0,0,0) and (1,1,1) have distance $\sqrt{3\cdot(0-1)^2} = \sqrt{3}$.
* so generally, for any $d$: $[0]^{1\times d}$ and $[1]^{1\times d}$ the distance separating them is: $\sqrt{d}$.

Note that at $d=9$ the distance will be already 3 times the distance the two initial points had in 1D !!

````python
import matplotlib.pyplot as plt
plt.style.use('fivethirtyeight')

import numpy as np
plt.figure(figsize=(12, 6))
plt.title(r"Euclidean distance between $[0]^{1 \times d}$ and $[1]^{1 \times d}$ for $d \in \{1, \ldots, 10\}$")
plt.xlabel('$d$')
plt.ylabel('$|| a - b ||_2$')
plt.plot(range(1, 11), np.sqrt(range(1, 11)))
````

Note that the increase of the distances between points has not the same gravity if the size of the space is kept the same; for instance, in 2D we could scale, with respect to the origin (which means to move the points in the space), the data such that all distances become some fixed multiple of the original distances. The difference here is that the increase of the distance does not come exactly from moving the points, but rather by augmenting the space in which they exist.

**b)** Imagine now that we generate data points uniformly at random in a unit $d$-hypercube, i.e. a cube in $d$ dimensions with unit edge length $e$. Then we put forward the following simple question connecting the proportion of volume of the space to the proportion of the data: 

*What is the proportion of the volume of the $d$-hypercube that we need to take in order to capture a proportion $r$ of the data?* 

<img src="hypercube.png" alt="Drawing" style="width: 300px;"/>

By definition, the volume of a unit $d$-hypercube is 

$V_d = e^d = 1$. 

With $e = 1$ we take the full volume that captures $r = 1$ (=100%) of the data: 

$e^d = r \Rightarrow  e = r^{1/d}$

Now, if we want to take a smaller cube inside the hypercube that contains the $r$ ratio of data, we can express its edge parametrized by the dimension $d$ and the ratio $r$:

$e_d(r) = r^{1/d}$

As we can see, as $d$ grows, any given ratio of data will occupy a bigger and bigger part of the data space. Just note that 10% of the data lies across the 0.1 of the unit length in 1D space, which means that they occupy 10% of the 1D space. However, the same 10% needs about... 80% of the volume in $d=10$ dimensions.

Intuitively, this means that the data points not only do they get farther away from each other, but they get also *squeezed* at the fringe of the space... (imagine points squeezed towards the sides of the cube) 


````python
plt.figure(figsize=(6, 6))
plt.title("% volume covered by $r$ uniform samples in a $d$-hypercube")
plt.xlabel('$d$')
plt.ylabel('% volume covered')

d_max = 11

plt.plot(range(1, d_max), [0.01**(1/n) for n in range(1, d_max)], label="r = 1%")
plt.plot(range(1, d_max), [0.1**(1/n) for n in range(1, d_max)], label="r = 10%")
plt.plot(range(1, d_max), [0.2**(1/n) for n in range(1, d_max)], label="r = 20%")
plt.plot(range(1, d_max), [0.33**(1/n) for n in range(1, d_max)], label="r = 33%")
plt.plot(range(1, d_max), [0.5**(1/n) for n in range(1, d_max)], label="r = 50%")
plt.plot(range(1, d_max), [0.8**(1/n) for n in range(1, d_max)], label="r = 80%")
plt.legend()
````
