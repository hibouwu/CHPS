**b)** 实现下面这个简单、确定性的代表点估计器，用于估计集合 $S$ 的代表点 $r$。该方法称为**合成原型（synthetic prototype）**：

步骤 1：$r \leftarrow \text{medoid}(S)$，即先取集合 $S$ 的中心样本点（medoid）。

步骤 2：$r \leftarrow$ 距离当前 $r$ 最近的 $x$% 个数据点的均值。

步骤 3：$r \leftarrow$ 距离当前 $r$ 最近的 $y$% 个数据点的均值。

……以此类推，其中 $x < y < ... < z$。也就是说，每一步都以上一步计算得到的代表点为中心，在一个逐步扩大的邻域内求均值。

实际使用中，通常两步就足够了，例如 $x=10$%、$y=80$%。研究结果对所选比例和步骤数量的敏感程度。由于该估计器会排除集合中的部分对象，因此它计算的是一种截尾均值（trimmed mean）。以下论文将这一方法用于构建簇原型：Kalogeratos 等，*Document clustering using synthetic cluster prototypes*，Data & Knowledge Engineering，2011（[参见算法 2](https://kalogeratos.com/files/MyPapers/Document-clustering-using-synthetic-cluster-prototypes-DKE2011-recompiled.pdf)；距离或相似度度量可以自行选择）。
